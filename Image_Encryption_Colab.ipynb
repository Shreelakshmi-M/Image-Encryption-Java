# 🔐 Image Encryption Using Pixel Value Transformation in Java

This notebook demonstrates a basic image encryption technique using Java. RGB pixel values are transformed using a user-defined key between 0 and 255.

## Step 1: Project Overview

The project reads an input image, processes every pixel, extracts the RGB components, applies modular arithmetic using the selected key, and saves the transformed image.

## Step 4: Java Encryption Program

The Java program processes every pixel. The RGB components are transformed using:

`R' = (R + Key) mod 256`  
`G' = (G + Key) mod 256`  
`B' = (B + Key) mod 256`

```python
%%writefile ImageEncoder.java
import java.awt.image.BufferedImage;
import java.io.File;
import javax.imageio.ImageIO;

public class ImageEncoder {

    public static void main(String[] args) {

        try {
            // Get encryption key from command-line argument
            int key = Integer.parseInt(args[0]);

            // Check whether the key is within the valid range
            if (key < 0 || key > 255) {
                System.out.println("Invalid key! Enter a value between 0 and 255.");
                return;
            }

            // Read the input image
            File inputFile = new File("input.png");

            if (!inputFile.exists()) {
                System.out.println("Input image not found!");
                return;
            }

            BufferedImage image = ImageIO.read(inputFile);

            int width = image.getWidth();
            int height = image.getHeight();

            // Process every pixel
            for (int y = 0; y < height; y++) {
                for (int x = 0; x < width; x++) {

                    int pixel = image.getRGB(x, y);

                    // Extract ARGB components
                    int alpha = (pixel >> 24) & 0xff;
                    int red = (pixel >> 16) & 0xff;
                    int green = (pixel >> 8) & 0xff;
                    int blue = pixel & 0xff;

                    // Transform RGB values using the encryption key
                    red = (red + key) % 256;
                    green = (green + key) % 256;
                    blue = (blue + key) % 256;

                    // Reconstruct the pixel
                    int newPixel =
                            (alpha << 24) |
                            (red << 16) |
                            (green << 8) |
                            blue;

                    image.setRGB(x, y, newPixel);
                }
            }

            // Save the encrypted image
            File outputFile = new File("encoded_image.png");
            ImageIO.write(image, "png", outputFile);

            System.out.println("Image encrypted successfully!");
            System.out.println("Encryption Key: " + key);
            System.out.println("Output File: encoded_image.png");

        } catch (NumberFormatException e) {
            System.out.println("Invalid input! Please enter a number between 0 and 255.");

        } catch (Exception e) {
            System.out.println("Error: " + e.getMessage());
        }
    }
}
```

The Java source code is compiled before execution.

## Step 6: Enter Encryption Key

The user can enter an encryption key between 0 and 255. The selected key is used to transform the RGB values of every pixel.

```python
key = int(input('Enter encryption key (0-255): '))

if key < 0 or key > 255:
    print('Invalid key! Please enter a value between 0 and 255.')
else:
    print('Encryption key accepted:', key)
```

## Step 7: Run the Java Encryption Program

The key entered by the user is passed to Java through a command-line argument.

```python
import subprocess

if 0 <= key <= 255:
    result = subprocess.run(
        ['java', 'ImageEncoder', str(key)],
        capture_output=True,
        text=True
    )
    print(result.stdout)
    if result.stderr:
        print(result.stderr)
```

## Step 8: Display the Encrypted Image

```python
from IPython.display import display
output_image = Image.open('encoded_image.png')
display(output_image)
```

## Step 9: Original vs Encrypted Image

The images are compared side by side. The image dimensions remain unchanged.

```python
original = Image.open('input.png')
encrypted = Image.open('encoded_image.png')

plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.imshow(original)
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(encrypted)
plt.title(f'Encrypted Image - Key {key}')
plt.axis('off')
plt.show()
```

## How the Pixel Transformation Works

For each pixel, Alpha, Red, Green and Blue are extracted. The key is added to each RGB component and the result is taken modulo 256. The Alpha component is preserved.
